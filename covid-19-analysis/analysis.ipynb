## Setup and data preview

```python
import pandas as pd
from IPython.display import display

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)
data_path = r'data/4. Covid_19_data.csv'
df = pd.read_csv(data_path)
print(f'Dataset shape: {df.shape}')
display(df.head())
```

## Q.4-1) Show Confirmed, Deaths and Recovered cases in each Region.

```python
region_totals = df.groupby('Region')[['Confirmed', 'Deaths', 'Recovered']].sum()
display(region_totals)
```

## Q.4-2) Remove records where Confirmed Cases is less than 10.

```python
before = len(df)
df = df[df['Confirmed'] >= 10].copy()
print(f'Removed {before - len(df)} rows; {len(df)} rows remain.')
display(df.head())
```

## Q.4-3) In which Region was the maximum number of Confirmed cases recorded?

```python
max_row = df.loc[df['Confirmed'].idxmax(), ['Region', 'Confirmed']]
display(max_row.to_frame('Value'))
```

## Q.4-4) In which Region was the minimum number of Deaths recorded?

```python
min_deaths = df['Deaths'].min()
result = df.loc[df['Deaths'] == min_deaths, ['Region', 'Deaths']]
print(f'Minimum deaths: {min_deaths}')
display(result)
```

## Q.4-5) Cases reported from India through 29 April 2020.

```python
india = df[df['Region'] == 'India'][['Confirmed', 'Deaths', 'Recovered']]
print('India totals:')
display(india.sum().to_frame('Cases'))
print('Source row(s):')
display(df[df['Region'] == 'India'])
```

## Q.4-6-A) Sort by Confirmed cases in ascending order.

```python
display(df.sort_values('Confirmed', ascending=True))
```

## Q.4-6-B) Sort by Recovered cases in descending order.

```python
display(df.sort_values('Recovered', ascending=False))
```
