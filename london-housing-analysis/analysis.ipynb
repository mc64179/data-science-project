## Setup and data preview

```python
import pandas as pd
from IPython.display import display

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)
data_path = r'data/5. London Housing Data.csv'
df = pd.read_csv(data_path)
print(f'Dataset shape: {df.shape}')
display(df.head())
```

## Q.5-1) Convert the Date column to Date-Time format.

```python
df['date'] = pd.to_datetime(df['date'])
print(df.dtypes)
```

## Q.5-2) Add year and month columns, with month as the second column.

```python
df['year'] = df['date'].dt.year
df.insert(1, 'month', df['date'].dt.month)
display(df.head())
```

## Q.5-3) Remove the year and month columns.

```python
df.drop(columns=['year', 'month'], inplace=True)
print(df.columns.tolist())
```

## Q.5-4) Show records where No. of Crimes is 0 and count them.

```python
zero_crime = df[df['no_of_crimes'] == 0]
print(f'Total records with zero crimes: {len(zero_crime)}')
display(zero_crime)
```

## Q.5-5) Maximum and minimum average_price per year in England.

```python
england = df[df['area'] == 'england'].copy()
england['year'] = england['date'].dt.year
price_by_year = england.groupby('year')['average_price'].agg(['max', 'min'])
display(price_by_year)
```

## Q.5-6) Maximum and minimum No. of Crimes recorded per area.

```python
crime_by_area = df.groupby('area')['no_of_crimes'].agg(['max', 'min'])
display(crime_by_area)
```

## Q.5-7) Count records in each area where average_price is less than 100000.

```python
counts = df[df['average_price'] < 100000]['area'].value_counts().sort_index()
display(counts.to_frame('Record count'))
```
