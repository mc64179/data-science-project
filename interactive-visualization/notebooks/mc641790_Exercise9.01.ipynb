## 1. Load and inspect the world population data

```python
import pandas as pd
from bokeh.io import output_notebook
output_notebook()
dataset = pd.read_csv('../data/world_population.csv', index_col=0)
dataset.head()
```

```python
years = [c for c in dataset.columns if not c[0].isalpha()]
de_vals = [float(dataset.loc['Germany', y]) for y in years]
ch_vals = [float(dataset.loc['Switzerland', y]) for y in years]
```

```python
from bokeh.plotting import figure, show

# Plot population density change in Germany
germany_plot = figure(
    title='Population Density of Germany',
    x_axis_label='Year',
    y_axis_label='Population Density',
    height=350
)

germany_plot.line(
    years,
    de_vals,
    line_width=2,
    legend_label='Germany'
)

show(germany_plot)
```

```python
from bokeh.plotting import figure, show
plot = figure(title='Population Density of Germany and Switzerland', x_axis_label='Year', y_axis_label='Population Density', height=350)
plot.line(years, de_vals, line_width=2, legend_label='Germany')
plot.line(years, ch_vals, line_width=2, color='orange', legend_label='Switzerland')
plot.scatter(years, ch_vals, size=5, color='orange', legend_label='Switzerland')
plot.legend.location='top_left'
show(plot)
```

```python
from bokeh.layouts import gridplot
plot_de=figure(title='Germany', x_axis_label='Year', y_axis_label='Population Density', height=300)
plot_ch=figure(title='Switzerland', x_axis_label='Year', y_axis_label='Population Density', height=300, x_range=plot_de.x_range, y_range=plot_de.y_range)
plot_de.line(years, de_vals, line_width=2)
plot_ch.line(years, ch_vals, line_width=2)

# Display the two plots horizontally
horizontal_grid = gridplot([[plot_de, plot_ch]])
show(horizontal_grid)

# Display the two plots vertically
vertical_grid = gridplot([[plot_de], [plot_ch]])
show(vertical_grid)
```
