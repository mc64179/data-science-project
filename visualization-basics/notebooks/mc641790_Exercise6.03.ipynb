# Exercise 6.03: Animal Correlation Scatter Plot

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline

# 读取真实数据集
data = pd.read_csv("anage_data.csv")

# 数据列名称
longevity = "Maximum longevity (yrs)"
mass = "Body mass (g)"

# 只保留质量和寿命都存在、且大于 0 的数据
data = data[
    np.isfinite(data[longevity])
    & np.isfinite(data[mass])
    & (data[longevity] > 0)
    & (data[mass] > 0)
]

# 按动物类别筛选
amphibia = data[data["Class"] == "Amphibia"]
aves = data[data["Class"] == "Aves"]
mammalia = data[data["Class"] == "Mammalia"]
reptilia = data[data["Class"] == "Reptilia"]

# 检查每一类的数据数量
print("Amphibia:", len(amphibia))
print("Aves:", len(aves))
print("Mammalia:", len(mammalia))
print("Reptilia:", len(reptilia))

# 创建图形
plt.figure(figsize=(10, 6), dpi=300)

# 绘制不同类别的散点
plt.scatter(
    amphibia[mass],
    amphibia[longevity],
    label="Amphibia"
)

plt.scatter(
    aves[mass],
    aves[longevity],
    label="Aves"
)

plt.scatter(
    mammalia[mass],
    mammalia[longevity],
    label="Mammalia"
)

plt.scatter(
    reptilia[mass],
    reptilia[longevity],
    label="Reptilia"
)

# 设置图例
plt.legend()

# 设置双对数坐标
ax = plt.gca()
ax.set_xscale("log")
ax.set_yscale("log")

# 设置坐标轴标签和标题
plt.xlabel("Body mass in grams")
plt.ylabel("Maximum longevity in years")
plt.title("Scatter plot on animal statistics")

plt.tight_layout()
plt.show()
```
