# Exercise 6.01: Simple Visualization
A line plot with circular markers.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
plt.style.use('seaborn-v0_8-whitegrid')
plt.figure(figsize=(6,4), dpi=200)
plt.plot([1,2,4,5], [1,3,4,3], '-o', color='#2563eb')
plt.xlabel('X'); plt.ylabel('Y'); plt.title('Simple line plot')
plt.tight_layout(); plt.savefig('Exercise6.01.png', bbox_inches='tight'); plt.show()
```
