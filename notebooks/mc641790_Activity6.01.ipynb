# Activity 6.01: Stock Trends
Line chart for five companies. The notebook uses generated representative data when CSV files are unavailable.

```python
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline

# 读取五个股票数据文件
google = pd.read_csv("GOOGL_data.csv")
facebook = pd.read_csv("FB_data.csv")
apple = pd.read_csv("AAPL_data.csv")
amazon = pd.read_csv("AMZN_data.csv")
microsoft = pd.read_csv("MSFT_data.csv")

# 将日期转换为日期格式
google["date"] = pd.to_datetime(google["date"])
facebook["date"] = pd.to_datetime(facebook["date"])
apple["date"] = pd.to_datetime(apple["date"])
amazon["date"] = pd.to_datetime(amazon["date"])
microsoft["date"] = pd.to_datetime(microsoft["date"])

# 创建图形
plt.figure(figsize=(12, 7), dpi=150)

# 绘制收盘价，顺序与课程资料一致
plt.plot(
    google["date"],
    google["close"],
    label="Google"
)

plt.plot(
    facebook["date"],
    facebook["close"],
    label="Facebook"
)

plt.plot(
    apple["date"],
    apple["close"],
    label="Apple"
)

plt.plot(
    amazon["date"],
    amazon["close"],
    label="Amazon"
)

plt.plot(
    microsoft["date"],
    microsoft["close"],
    label="Microsoft"
)

# 添加标题、坐标轴标签和图例
plt.title("Stock trend")
plt.xlabel("Date")
plt.ylabel("Closing price in $")
plt.legend()

# 添加网格
plt.grid(True)

# 自动调整日期标签
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
