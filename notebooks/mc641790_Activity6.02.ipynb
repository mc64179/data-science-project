# Activity 6.02: Movie Score Comparison

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline

# 读取电影评分数据
df = pd.read_csv("movie_scores.csv", index_col=0)

movies = df["MovieTitle"].tolist()
tomatometer = df["Tomatometer"].tolist()
audience = df["AudienceScore"].tolist()

x = np.arange(len(movies))
width = 0.36

fig, ax = plt.subplots(figsize=(10, 6), dpi=150)

ax.bar(
    x - width / 2,
    tomatometer,
    width,
    label="Tomatometer",
    color="#2563eb"
)

ax.bar(
    x + width / 2,
    audience,
    width,
    label="Audience Score",
    color="#f97316"
)

ax.set_xlabel("Movie")
ax.set_ylabel("Score (%)")
ax.set_title("Movie Score Comparison")

ax.set_xticks(x)
ax.set_xticklabels(movies, rotation=20, ha="right")

ax.set_ylim(0, 105)
ax.set_yticks(np.arange(0, 101, 20))
ax.set_yticks(np.arange(0, 101, 5), minor=True)

ax.yaxis.grid(which="major", linestyle="-", alpha=0.6)
ax.yaxis.grid(which="minor", linestyle="--", alpha=0.3)

ax.legend()
plt.tight_layout()
plt.show()
```
