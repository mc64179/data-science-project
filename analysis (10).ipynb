## Setup and data preview

```python
import pandas as pd
from IPython.display import display

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)
data_path = r'data/2. Cars Data.csv'
df = pd.read_csv(data_path)
print(f'Dataset shape: {df.shape}')
display(df.head())
```

## Q.2-1) Find Null Values and fill numeric nulls with each column's mean.

```python
print('Null values before cleaning:')
display(df.isnull().sum().to_frame('Null Values'))
for col in df.select_dtypes(include='number').columns:
    df[col] = df[col].fillna(df[col].mean())
print('Null values after cleaning:')
display(df.isnull().sum().to_frame('Null Values'))
```

## Q.2-2) List each Make and its occurrence count.

```python
make_counts = df['Make'].value_counts()
print(f'Number of different Makes: {df["Make"].nunique()}')
display(make_counts.to_frame('Count'))
```

## Q.2-3) Show all records where Origin is Asia or Europe.

```python
result = df[df['Origin'].isin(['Asia', 'Europe'])]
print(f'Total matching records: {len(result)}')
display(result)
```

## Q.2-4) Remove all records where Weight is above 4000.

```python
before = len(df)
df = df[df['Weight'] <= 4000].copy()
print(f'Removed {before - len(df)} rows; {len(df)} rows remain.')
display(df.head())
```

## Q.2-5) Increase all MPG_City values by 3.

```python
df['MPG_City'] = df['MPG_City'].apply(lambda value: value + 3)
print(df['MPG_City'].describe())
display(df[['Make', 'Model', 'MPG_City']].head(10))
```
