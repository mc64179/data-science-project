## Setup and data preview

```python
import pandas as pd
from IPython.display import display

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)
data_path = r'data/3. Police Data.csv'
df = pd.read_csv(data_path)
print(f'Dataset shape: {df.shape}')
display(df.head())
```

## Q.3-1) Remove the column that contains only missing values.

```python
all_null_columns = df.columns[df.isnull().all()].tolist()
print('All-null columns:', all_null_columns)
df.drop(columns=all_null_columns, inplace=True)
print('Remaining columns:', df.columns.tolist())
```

## Q.3-2) For Speeding, were Men or Women stopped more often?

```python
speeding_gender = df.loc[df['violation'] == 'Speeding', 'driver_gender'].value_counts()
display(speeding_gender.to_frame('Speeding stops'))
print(f"Answer: {speeding_gender.idxmax()} drivers were stopped more often for speeding.")
```

## Q.3-3) Does gender affect who gets searched during a stop?

```python
search_counts = pd.crosstab(df['driver_gender'], df['search_conducted'], margins=True)
search_rates = df.groupby('driver_gender')['search_conducted'].mean().mul(100)
display(search_counts)
display(search_rates.to_frame('Search rate (%)'))
print('The search rates differ by gender, so gender is associated with the likelihood of a search in this dataset.')
```

## Q.3-4) What is the mean stop_duration?

```python
duration_minutes = df['stop_duration'].map({'0-15 Min': 7.5, '16-30 Min': 23, '30+ Min': 45})
print('Duration categories:')
display(df['stop_duration'].value_counts(dropna=False).to_frame('Count'))
print(f'Estimated mean stop duration: {duration_minutes.mean():.2f} minutes')
```

## Q.3-5) Compare age distributions for each violation.

```python
age_by_violation = df.groupby('violation')['driver_age'].describe()
display(age_by_violation)
```
