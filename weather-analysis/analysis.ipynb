## Setup and data preview

```python
import pandas as pd
from IPython.display import display

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)
data_path = r'data/1. Weather Data.csv'
df = pd.read_csv(data_path)
print(f'Dataset shape: {df.shape}')
display(df.head())
```

## Q.1-1) Find all the unique 'Wind Speed' values in the data.

```python
wind_speeds = sorted(df['Wind Speed_km/h'].unique())
print(f'Number of unique values: {len(wind_speeds)}')
print(wind_speeds)
```

## Q.1-2) Find the number of times when the weather is exactly Clear.

```python
clear_count = (df['Weather'] == 'Clear').sum()
print(f'Weather was exactly Clear {clear_count} times.')
```

## Q.1-3) Find the number of times when the Wind Speed was exactly 4 km/h.

```python
wind_4_count = (df['Wind Speed_km/h'] == 4).sum()
print(f'Wind Speed was exactly 4 km/h {wind_4_count} times.')
```

## Q.1-4) Find all Null Values in the data.

```python
null_counts = df.isnull().sum()
display(null_counts.to_frame('Null Values'))
print(f'Total null values: {null_counts.sum()}')
```

## Q.1-5) Rename 'Weather' to 'Weather Condition'.

```python
df.rename(columns={'Weather': 'Weather Condition'}, inplace=True)
print(df.columns.tolist())
```

## Q.1-6) What is the mean Visibility?

```python
print(f"Mean Visibility: {df['Visibility_km'].mean():.6f} km")
```

## Q.1-7) What is the Standard Deviation of Pressure?

```python
print(f"Pressure standard deviation: {df['Press_kPa'].std():.6f} kPa")
```

## Q.1-8) What is the Variance of Relative Humidity?

```python
print(f"Relative Humidity variance: {df['Rel Hum_%'].var():.6f}")
```

## Q.1-9) Find all instances when Snow was recorded.

```python
snow = df[df['Weather Condition'].str.contains('Snow', case=False, na=False)]
print(f'Total matching records: {len(snow)}')
display(snow)
```

## Q.1-10) Find instances when Wind Speed is above 24 and Visibility is 25.

```python
result = df[(df['Wind Speed_km/h'] > 24) & (df['Visibility_km'] == 25)]
print(f'Total matching records: {len(result)}')
display(result)
```

## Q.1-11) Mean value of each numeric column by Weather Condition.

```python
display(df.groupby('Weather Condition').mean(numeric_only=True))
```

## Q.1-12) Minimum and Maximum of each column by Weather Condition.

```python
summary = df.groupby('Weather Condition').agg(['min', 'max'])
display(summary)
```

## Q.1-13) Show all records where Weather Condition is Fog.

```python
fog = df[df['Weather Condition'] == 'Fog']
print(f'Total Fog records: {len(fog)}')
display(fog)
```

## Q.1-14) Find instances when Weather is Clear or Visibility is above 40.

```python
result = df[(df['Weather Condition'] == 'Clear') | (df['Visibility_km'] > 40)]
print(f'Total matching records: {len(result)}')
display(result)
```

## Q.1-15) Find instances when (Weather is Clear and Relative Humidity > 50) or Visibility > 40.

```python
result = df[((df['Weather Condition'] == 'Clear') & (df['Rel Hum_%'] > 50)) | (df['Visibility_km'] > 40)]
print(f'Total matching records: {len(result)}')
display(result)
```
